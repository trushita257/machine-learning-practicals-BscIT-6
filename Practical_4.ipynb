```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/content/covid_toy.csv")
df.head()
```

```python
print(df.cough.value_counts())
print()
print()

print(df.city.value_counts())
```

```python
'''
age-numeric column
gender-Nominal categorical column(One-Hot encoding)
fever-Numeric column.Has missing values.Will apply SimpleImputer
cough-Ordinal categorical column(Oridinal encoding)
city-Nominal categorical column
has_covid-Nominal categorical column
'''
```

```python
df.isnull().sum()
```

```python
x=df.iloc[:,0:5]
y=df.drop(x,axis='columns')

print(x.head())

print()
print()
print()

print(y.head())
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
print(x_train.head())
print()
print()
print()
print(y_train.head())
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder
```

```python
si=SimpleImputer()
le=LabelEncoder()
```

```python
y_train.has_covid=le.fit_transform(y_train.has_covid)
y_test.has_covid=le.transform(y_test['has_covid'])
```

```python
x_train['fever']=si.fit_transform(x_train[['fever']])
x_test['fever']=si.transform(x_test[['fever']])
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe=OrdinalEncoder()

x_train['cough']=oe.fit_transform(x_train[['cough']])
x_test['cough']=oe.transform(x_test[['cough']])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder(drop='first',sparse_output=False)
```

```python
x_train_gender_city=ohe.fit_transform(x_train[['gender','city']])
x_test_gender_city=ohe.transform(x_test[['gender','city']])

print(x_train_gender_city.shape)
print(x_test_gender_city.shape)
```

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer=[('trans1',SimpleImputer(),['fever']),
             ('trans2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
             ('trans3',OneHotEncoder(drop='first',sparse_output=False),['gender','city'])
              ]
```

```python
ct=ColumnTransformer(transformers=transformer,
                     remainder='passthrough').set_output(transform='pandas')
```

```python
x_train_trans=ct.fit_transform(x_train)
```
