```python
0.import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler,MinMaxScaler,OneHotEncoder
#from sklearn.compose import ColumnTransform
#from sklearn.pipeline import Pipeline
#from sklearn.impute import SimpleImputer
from sklearn.feature_selection importSelectKBest,f_regression,f_classif
#from sklearn.decomposition import PCA
```

```python
df=pd.read_csv("/content/auto-mpg (2).csv")

display(df.head())
```

```python
print(df.dtypes)
```

```python
numeric_columns=df.select_dtypes(include=np.number).columns

print("Numeric Columns:")
print(numeric_columns)
```

```python
categorical_columns=df.select_dtypes(include="object").columns

print("Categorical Columns:")
print(categorical_columns)
```

```python
df["origin"]=df['origin'].astype("object")
```

```python
print(df.isnull().sum())
```

```python
df=df.replace("?",np.nan)
```

```python
print(df.isnull().sum())
```

```python
df["horsepower"]=pd.to_numeric(df["horsepower"])
```

```python
df["horsepower"]=df["horsepower"].fillna(df["horsepower"].mean())
```

```python
print(df.isnull().sum())
```

```python
print(df.isnull().sum())
```

```python
scaler=MinMaxScaler().set_output(transform="pandas")

scaled_data=scaler.fit_transform(df[numeric_columns])

scaled_data.head()
```

```python
categorical_columns=["origin","car name"]
```

```python
encoder=OneHotEncoder(drop="first").set_output(transform="pandas")

encoded_data=encoder.fit_transform(df[categorical_columns])

encoded_data.head()
```

```python
X = df.drop("mpg, axis=1")
y = df["mpg"]
```

```python
X_numeric = X.select_dtypes(include=np.number)

selector = SelectKBest(
    score_func=f_regression,
    k=5
).set_output(transform="pandas")

X_selected=selector.fit_transform(X_numeric, y)

print(X_selected.head())

```
